```python
from pathlib import Path
import sys
base_path = next(
    p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
    if (p / "CODE").is_dir() and (p / "DATA").is_dir()
)
sys.path.insert(0, str(Path(base_path) / 'CODE'))
import re

from utils.surface import plot_parcelwise_with_roi_surface_direct
```

```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
output_path = f"{base_path}/FIGURES/supplementary/Fig_S6"
os.makedirs(output_path, exist_ok=True)

task_colors = {
    "Rest": '#ff5e69',   
    "0Back": '#86ab00',
    "1Back": '#03b6ce', 
    "2Back": '#8491ff'  
}

task_names = ["Rest", "0Back", "1Back", "2Back"]

data_path = f"{base_path}/RESULTS/replication_dataset/anova/anova_results_parcelwise_FIN_D4_modwt_1_5_padding_0_decimate_0_residuals_rest_no_outliers.xlsx"
f_values_df = pd.read_excel(data_path)
```

```python
from utils.contrasts import (
    parse_pairwise_key, make_task_diverging_cmap, get_pairwise_keys_from_df,
)

pairwise_keys = get_pairwise_keys_from_df(f_values_df)
print("Contrasts found:", pairwise_keys)

for key in pairwise_keys:
    task1, task2 = parse_pairwise_key(key, task_names)

    tvals = pd.to_numeric(f_values_df[f"t_{key}"], errors="coerce").to_numpy(dtype=float)
    sig = f_values_df[f"sig_{key}_fdr"].fillna(False).to_numpy(dtype=bool)

    tvals_plot = np.where(sig, tvals, 0.0)

    cmap = make_task_diverging_cmap(task1, task2, task_colors)

    n_sig = int(np.sum(sig & ~np.isnan(tvals)))
    print(f"Plotting {key}: {task1} vs {task2} | significant parcels = {n_sig}")

    plot_parcelwise_with_roi_surface_direct(
        tvals_plot,
        lh_annot_path=f"{base_path}/DATA/templates/lh.Schaefer2018_100Parcels_7Networks_order.annot",
        rh_annot_path=f"{base_path}/DATA/templates/rh.Schaefer2018_100Parcels_7Networks_order.annot",
        roi_img_path=None,
        vmin=-9,
        vmax=9,
        img_name=f"{output_path}/{task1}_vs_{task2}.png",
        show=False,
        inflate_fraction=0.70,
        colormap=cmap
    )
```
